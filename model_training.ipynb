```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Load the dataset
file_path = '/content/EPM 4.csv'
data = pd.read_csv(file_path)

# Calculate distance between two consecutive frames
def calculate_distance(x1, y1, x2, y2):
    return np.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Initialize new columns for distance and speed
data['distance'] = 0.0
data['speed'] = 0.0

# Loop over each row and calculate distance and speed
for i in range(1, len(data)):
    # Calculate distance between frame i and frame i-1
    data.loc[i, 'distance'] = calculate_distance(data.loc[i-1, 'bodypart1_x'], data.loc[i-1, 'bodypart1_y'],
                                                 data.loc[i, 'bodypart1_x'], data.loc[i, 'bodypart1_y'])

    # Calculate speed (distance / time), assuming a constant time interval (e.g., 1 second)
    data.loc[i, 'speed'] = data.loc[i, 'distance']  # Replace 1 with the actual time interval if known

# Add new features based on the differences in body part positions
data['bodypart1_diff'] = data['bodypart1_x'] - data['bodypart1_y']
data['bodypart2_diff'] = data['bodypart2_x'] - data['bodypart2_y']
data['bodypart3_diff'] = data['bodypart3_x'] - data['bodypart3_y']

# Define a function to categorize behavior based on speed and distance
def categorize_behavior(speed, distance):
    if speed < 1:  # Adjust the threshold as needed
        return 'stopping'
    elif distance > 10 and speed > 5:  # Adjust conditions based on observations
        return 'exploring'
    else:
        return 'moving'

# Apply the behavior categorization
data['behavior'] = data.apply(lambda row: categorize_behavior(row['speed'], row['distance']), axis=1)

# Encode target variable (behavior) into numerical format
data['behavior'] = data['behavior'].astype('category').cat.codes

# Features: distance, speed, and new features (bodypart differences)
X = data[['distance', 'speed', 'bodypart1_diff', 'bodypart2_diff', 'bodypart3_diff']]
y = data['behavior']  # Target: behavior labels

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Feature scaling (important for neural networks)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Build the Neural Network model
model = Sequential()

# Add input layer and two hidden layers
model.add(Dense(64, input_dim=X_train_scaled.shape[1], activation='relu'))
model.add(Dense(32, activation='relu'))

# Add output layer (use softmax for multi-class classification)
model.add(Dense(3, activation='softmax'))

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Train the model
history = model.fit(X_train_scaled, y_train, epochs=100, batch_size=32, validation_data=(X_test_scaled, y_test))

# Evaluate the model on the test set
test_loss, test_accuracy = model.evaluate(X_test_scaled, y_test)

print(f'\nNeural Network Test Accuracy: {test_accuracy}')

# Predict on the test set
y_pred = model.predict(X_test_scaled)
y_pred_classes = np.argmax(y_pred, axis=1)

# Evaluate the Neural Network model
print("\nNeural Network Classification Report:")
print(classification_report(y_test, y_pred_classes))
print("Accuracy:", accuracy_score(y_test, y_pred_classes))
```

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical

# Load the cleaned dataset
file_path = '/content/cleaned_VITE1002_cleaned.csv'  # Adjust this if necessary
cleaned_data = pd.read_csv(file_path)

# Feature Engineering: Calculate distance and speed
def calculate_distance(x1, y1, x2, y2):
    return np.sqrt((x2 - x1)**2 + (y2 - y1)**2)

cleaned_data['distance'] = 0.0
cleaned_data['speed'] = 0.0

# Loop to calculate distance and speed
for i in range(1, len(cleaned_data)):
    cleaned_data.loc[i, 'distance'] = calculate_distance(
        cleaned_data.loc[i-1, 'bodypart1_x'], cleaned_data.loc[i-1, 'bodypart1_y'],
        cleaned_data.loc[i, 'bodypart1_x'], cleaned_data.loc[i, 'bodypart1_y']
    )
    cleaned_data.loc[i, 'speed'] = cleaned_data.loc[i, 'distance']

# Categorize behavior based on speed and distance
def categorize_behavior(speed, distance):
    if speed < 1:
        return 0  # 'stopping'
    elif distance > 10 and speed > 5:
        return 2  # 'exploring'
    else:
        return 1  # 'moving'

cleaned_data['behavior'] = cleaned_data.apply(lambda row: categorize_behavior(row['speed'], row['distance']), axis=1)

# Split the data into features (X) and labels (y)
X = cleaned_data[['distance', 'speed']].values
y = cleaned_data['behavior'].values

# One-hot encode the labels for neural network training
y_categorical = to_categorical(y, num_classes=3)

# Define K-fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Function to create the model
def create_model():
    model = Sequential()
    model.add(Dense(64, input_dim=2, activation='relu'))  # Input layer with 2 features (distance and speed)
    model.add(Dense(32, activation='relu'))               # Hidden layer
    model.add(Dense(3, activation='softmax'))             # Output layer for 3 classes (stopping, moving, exploring)

    # Compile the model
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model

# Store accuracy and loss for each fold
fold_no = 1
acc_per_fold = []
loss_per_fold = []

for train_index, test_index in kf.split(X):
    # Split data for this fold
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y_categorical[train_index], y_categorical[test_index]

    # Create a new instance of the model
    model = create_model()

    # Train the model
    history = model.fit(X_train, y_train, epochs=30, batch_size=10, verbose=0, validation_data=(X_test, y_test))

    # Evaluate the model
    scores = model.evaluate(X_test, y_test, verbose=0)
    print(f"Fold {fold_no} - Loss: {scores[0]}, Accuracy: {scores[1]*100:.2f}%")

    acc_per_fold.append(scores[1] * 100)
    loss_per_fold.append(scores[0])

    fold_no += 1

# Average scores across all folds
print(f'\nAverage accuracy across all folds: {np.mean(acc_per_fold):.2f}%')
print(f'Average loss across all folds: {np.mean(loss_per_fold):.4f}')

# Confusion Matrix
y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = np.argmax(y_test, axis=1)

cm = confusion_matrix(y_true_classes, y_pred_classes)
ConfusionMatrixDisplay(cm, display_labels=['stopping', 'moving', 'exploring']).plot()
plt.title('Confusion Matrix (Last Fold)')
plt.show()

# Visualize training accuracy and loss for the last fold
plt.plot(history.history['accuracy'], label='train accuracy')
plt.plot(history.history['val_accuracy'], label='val accuracy')
plt.title('Training and Validation Accuracy (Last Fold)')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

plt.plot(history.history['loss'], label='train loss')
plt.plot(history.history['val_loss'], label='val loss')
plt.title('Training and Validation Loss (Last Fold)')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()

# Visualize Speed Distribution
plt.hist(cleaned_data['speed'], bins=30, color='c', edgecolor='k')
plt.xlabel('Speed')
plt.ylabel('Frequency')
plt.title('Distribution of Speed')
plt.show()

# Visualize Distance Distribution
plt.hist(cleaned_data['distance'], bins=30, color='g', edgecolor='k')
plt.xlabel('Distance')
plt.ylabel('Frequency')
plt.title('Distribution of Distance')
plt.show()
```
